```python
import numpy as np
# Creation of data
x=np.array([0.70,0.80,0.65])
print("\nInput data as a row vector")
print(x)
print(type(x))
```

```python
x=np.array([[.70],[.80],[.65]])
print("Input data as a column vector")
print(x)
print(f"The size of input data is{x.shape}")
```

```python
# Wieght and bias creation
W1=np.array([[.5,.2,-.1],[.8,.4,.3],[-.3,.6,.5]])
print("The weight matrix is...")
print(W1)
print("The size of the weight matrix is..")
print(W1.shape)
# Bias creation
b1=np.array([[.1],[.2],[.1]])
print('The bias vector is...')
print(b1)
print("The size of the bias vector is...")
print(b1.shape)
```

```python
# Activation Functions
def relu(z):
    return np.maximum(0,z)
def sigmoid(z):
    return 1/(1+np.exp(-z))

    
```

```python
# Forward Propagation
# Hidden layer 1
z1=W1.T@x+b1
print(z1)
```

```python
# Activation Function
a1=relu(z1)
print("Hidden layer pre-activation")
print(a1)
```

```python
# Hidden layer 2
W2=np.array([[.7,.5,-.2]])
b2=np.array([.1])
print("The weight matrix is...")
print(W2)
print("The size of the weight matrix is...")
print(W2.shape)
print("The bias is...")
print(b2)
```

```python
# Hidden layer 2 Output 
z2=W2@a1+b2

print(z2)
```

```python
y_hat=sigmoid(z2)
print(y_hat)
```

```python
probability=float(y_hat[0,0])
print(probability)
```

```python
if probability>0.5:
    print("Prediction:Good")
else :
    print("Prediction:Poor")
```
